```python
import numba
from numba import cuda

print("Numba version:", numba.__version__)
print("CUDA available:", cuda.is_available())
```

```python
from numba import cuda

cuda.detect()

device = cuda.select_device(0)

name = device.name
if isinstance(name, bytes):      # một số bản Numba trả về bytes, ví dụ b'Tesla T4'
    name = name.decode()

print("ID:", device.id)
print("Name:", name)
print("Compute capability:", device.compute_capability)
```

```python
from numba import cuda

CORES_PER_SM = {
    (6, 0): 64,    # Pascal - Tesla P100
    (6, 1): 128,   # Pascal - GTX 1080, GTX 1070...
    (7, 0): 64,    # Volta  - Tesla V100
    (7, 5): 64,    # Turing - Tesla T4, RTX 2080 Ti  <- GPU của Colab
    (8, 0): 64,    # Ampere - A100
    (8, 6): 128,   # Ampere - RTX 3090, RTX 3080...
}


device = cuda.select_device(0)

cc = tuple(device.compute_capability)

sm_count = device.MULTIPROCESSOR_COUNT

print("SM count:", sm_count)

print("Cores per SM:", CORES_PER_SM[cc])

print("Total cores:", sm_count * CORES_PER_SM[cc])
```

```python
from numba import cuda

context = cuda.current_context()

free_bytes, total_bytes = context.get_memory_info()

total_mb = total_bytes / (1024 ** 2)
total_gb = total_bytes / (1024 ** 3)
free_mb = free_bytes / (1024 ** 2)

print("Total memory:", total_bytes, "bytes")
print(f"Total memory: {total_mb:.2f} MB ({total_gb:.2f} GB)")
print(f"Free memory:  {free_mb:.2f} MB")
```
